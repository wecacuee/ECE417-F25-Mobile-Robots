### Rigid Body Transformation Concepts

A **Rigid body** is defined as a set of points where the distance between any two points remains constant after a transformation.

The transformation involves a **Rotation** ($R$) and a **Translation** ($t$). All points in the rigid body rotate and translate by the same amount.

The transformation of a point $\underline{p}_J$ in the source frame $J$ to a point $\underline{p}_W$ in the destination frame $W$ is given by:

$$
\underline{p}_W = {}^W R_J \underline{p}_J + {}^W \underline{t}_J \tag{1}
$$  

* $\underline{p}_W$: Point in the World/Destination frame ($W$)
* $\underline{p}_J$: Point in the Source frame ($J$)
* ${}^W R_J$: **Rotation matrix** from frame $J$ to $W$
* ${}^W \underline{t}_J$: **Translation vector** from frame $J$ to $W$

In 2D, this equation expands to:

$$\begin{pmatrix} x_W \\ y_W \end{pmatrix} = \begin{pmatrix} r_{11} & r_{12} \\ r_{21} & r_{22} \end{pmatrix} \begin{pmatrix} x_J \\ y_J \end{pmatrix} + \begin{pmatrix} t_1 \\ t_2 \end{pmatrix}
$$

### 2D Rotation Derivation

The notes derive the transformation for a 2D rotation of a point $\underline{p} = \begin{pmatrix} x \\ y \end{pmatrix}$ rotated by an angle $\theta$ to a new point $\underline{p}' = \begin{pmatrix} x' \\ y' \end{pmatrix}$.

Given:

* $x = r \cos\alpha$
* $y = r \sin\alpha$
* $x' = r \cos(\theta+\alpha)$
* $y' = r \sin(\theta+\alpha)$

Using the angle addition formulas ($\cos(A+B) = \cos A \cos B - \sin A \sin B$ and $\sin(A+B) = \sin A \cos B + \cos A \sin B$):

$$x' = r \cos\theta \cos\alpha - r \sin\theta \sin\alpha
$$Substituting $x$ and $y$:

$$
x' = x \cos\theta - y \sin\theta
$$

$$
y' = r \sin\theta \cos\alpha + r \cos\theta \sin\alpha
$$

Substituting $x$ and $y$:

$$y' = x \sin\theta + y \cos\theta  
$$

In matrix form, the **2D Rotation Matrix** $R(\theta)$ is:

$$
\underline{p}' = R(\theta)\underline{p}
$$

$$
\begin{pmatrix} x' \\ y' \end{pmatrix} = \begin{pmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix}
$$

### Homogeneous Coordinates and Transformation Matrix

To represent rotation and translation (Equation 1) as a single matrix multiplication, **homogeneous coordinates** are used. A 2D point $\underline{p} = \begin{pmatrix} x \\ y \end{pmatrix}$ is extended to $\begin{pmatrix} x \\ y \\ 1 \end{pmatrix}$.

The **Transformation Matrix** $T$ in homogeneous coordinates for a transformation from $J$ to $W$ is:

$$
{}^W T_J = \begin{pmatrix} {}^W R_J & {}^W \underline{t}_J \\ \underline{0}^T & 1 \end{pmatrix}
$$

Where $\underline{0}^T$ is a row vector of zeros (e.g., $\begin{pmatrix} 0 & 0 \end{pmatrix}$ for 2D).

The transformation becomes:

$$\begin{pmatrix} \underline{p}_W \\ 1 \end{pmatrix} = {}^W T_J \begin{pmatrix} \underline{p}_J \\ 1 \end{pmatrix}
$$

### Properties of a Rotation Matrix

A **Rotation Matrix** ($R$) is a type of **orthonormal matrix**. An orthonormal matrix $U$ has columns (or rows) that are orthonormal, meaning:

* $\underline{u}_i^T \underline{u}_i = 1$ (Unit length)
* $\underline{u}_i^T \underline{u}_j = 0$ for $i \ne j$ (Orthogonal)

#### 1\. Inverse and Transpose

For any orthonormal matrix $U$, its inverse is equal to its transpose:

$$U^{-1} = U^T
$$Therefore, for a rotation matrix $R$:

$$
R^{-1} = R^T
$$This property is geometrically intuitive: transforming a point back from $P'$ to $P$ (reversing a rotation by $\theta$) is equivalent to a rotation by $-\theta$.

$$R^{-1}(\theta) = R(-\theta) = \begin{pmatrix} \cos(-\theta) & -\sin(-\theta) \\ \sin(-\theta) & \cos(-\theta) \end{pmatrix} = \begin{pmatrix} \cos\theta & \sin\theta \\ -\sin\theta & \cos\theta \end{pmatrix}
$$

Which is $R^T$:

$$
R^T = \begin{pmatrix} \cos\theta & \sin\theta \\ -\sin\theta & \cos\theta \end{pmatrix}
$$Also, $R R^T = R^T R = I$ (Identity matrix).

#### 2\. Determinant

For an orthonormal matrix $U$, the determinant must be $\det(U) \in \{-1, +1\}$.

A pure **Rotation matrix** $R$ has a determinant of positive one:

$$\det(R) = 1
$$For the 2D rotation matrix:

$$
\det\left(\begin{pmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{pmatrix}\right) = \cos^2\theta - (-\sin^2\theta) = \cos^2\theta + \sin^2\theta = 1
$$

If an orthonormal matrix has $\det(U) = -1$, it represents a **Reflection matrix**. Thus, not all orthonormal matrices are rotation matrices.

### Inverse Transformation

The notes also demonstrate how to find the inverse transformation, ${}^J T_W = ({}^W T_J)^{-1}$.

Starting from the point transformation:

$$\underline{p}_W = {}^W R_J \underline{p}_J + {}^W \underline{t}_J
$$1.  Isolate the rotational term:

$$ \underline{p}_W - {}^W \underline{t}_J = {}^W R_J \underline{p}_J
$$


2.  Left-multiply by the transpose of the rotation matrix, ${}^W R_J^T$, which is the inverse ${}^J R_W$:

$${}^W R_J^T (\underline{p}_W - {}^W \underline{t}_J) = \left( {}^W R_J^T {}^W R_J \right) \underline{p}_J
$$

Since ${}^W R_J^T {}^W R_J = I$:

$${}^W R_J^T \underline{p}_W - {}^W R_J^T {}^W \underline{t}_J = \underline{p}_J
$$
This shows the inverse transformation is:

$$
\underline{p}_J = \underbrace{{}^W R_J^T}_{{}^J R_W} \underline{p}_W + \underbrace{\left( -{}^W R_J^T {}^W \underline{t}_J \right)}_{{}^J \underline{t}_W}
$$

The inverse transformation matrix ${}^J T_W$ in homogeneous coordinates is:

$${}^J T_W = \begin{pmatrix} {}^W R_J^T & -{}^W R_J^T {}^W \underline{t}_J \\ \underline{0}^T & 1 \end{pmatrix} = \begin{pmatrix} {}^J\underline{R}_W & {}^J\underline{t}_W \\ \underline{0}^T & 1 \end{pmatrix}
$$
