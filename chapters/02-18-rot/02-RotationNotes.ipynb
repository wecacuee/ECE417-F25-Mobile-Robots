# 3D Rotations and Transformations

## 3D Rotations

**3D Rotations** are fundamental transformations in space.

### 2D Rotation around Z-axis

The rotation of a point $(x, y)$ in the $XY$-plane (around the $Z$-axis) by an angle $\phi$ (or $\theta$) is given by:

$$
\begin{bmatrix}
x'\\
y'\\
z'
\end{bmatrix}
= 
\begin{bmatrix}
\cos\phi & -\sin\phi & 0\\
\sin\phi & \cos\phi & 0\\
0 & 0 & 1
\end{bmatrix}
\begin{bmatrix}
x\\
y\\
z
\end{bmatrix}
$$

## 3D Axis-Aligned Rotation Matrices

### Rotation around X-axis ($R_x(\theta_x)$)


$$ 
\begin{bmatrix}
x'\\
y'\\
z'
\end{bmatrix}
= 
\begin{bmatrix}
1 & 0 & 0\\
0 & \cos\theta_x & -\sin\theta_x\\
0 & \sin\theta_x & \cos\theta_x
\end{bmatrix}
\begin{bmatrix}
x\\
y\\
z
\end{bmatrix}
$$ 

### Rotation around Y-axis ($R_y(\theta_y)$)

$$ 
\begin{bmatrix}
x'\\
y'\\
z'
\end{bmatrix}
= 
\begin{bmatrix}
\cos\theta_y & 0 & \sin\theta_y\\
0 & 1 & 0\\
-\sin\theta_y & 0 & \cos\theta_y
\end{bmatrix}
\begin{bmatrix}
x\\
y\\
z
\end{bmatrix}
$$ 


## Right-Hand Coordinate Frame

The rotations typically follow the **Right-hand coordinate frame** convention

To determine the positive direction of rotation, curl your fingers from the first axis to the second; your thumb points in the direction of the third axis (the rotation axis).

* $X \rightarrow Y$: positive rotation is around $Z$
* $Y \rightarrow Z$: positive rotation is around $X$


## Euler Angles and Decomposition

**All 3D rotations** can be **decomposed into 3 axis-aligned rotations**

These three angles of rotation are called **Euler angles**.

A common set of Euler angles is **Roll-Pitch-Yaw** (RPY):
* $\theta_x = \text{roll}$ 
* $\theta_y = \text{pitch}$ 
* $\theta_z = \text{yaw}$ 

The total rotation $R$ is a product of the individual rotations. For the Roll-Pitch-Yaw sequence ($X-Y-Z$):

$$
R = R_x(\theta_z) R_y(\theta_y) R_z(\theta_z)
$$

**Important:** The **Application of Rotation matrices is from right to left**


## Combination of Euler Angles

There are **12 possible** combinations of Euler angles (rotations).

This arises from two key choices when chaining rotations:

1.  **Axis Permutations:** There are $3! = 6$ possibilities for the order of axes (e.g., $X-Y-Z, Y-Z-X, Z-Y-Z$).
2.  **Tait-Bryant Angles:** You only use two axis but three rotations (e.g., $X-Y-X, X-Z-X$).


## Composition of Rotations / Transformations

### Rotation Composition
The rotation from frame $C$ to frame $W$ via an intermediate frame $S$ is a matrix product:
$$ 
\mathbf{^wR_c} = \mathbf{^wR_s} \mathbf{^sR_c}
$$ 

The composition of two rotation matrices, $R_1 R_2$, is also orthonormal with a determinant of 1.
$$
\det(R_1 R_2) = \det(R_1) \det(R_2) = 1 \cdot 1 = 1
$$
The inverse of a composed rotation is: $(R_1 R_2)^{-1} = R_2^{-1} R_1^{-1} = R_2^T R_1^T$

### Homogeneous Transformation Composition
For homogeneous transformations $\mathbf{T}$ (which include both rotation and translation):
$$ 
\mathbf{^wT_c} = \mathbf{^wT_s} \mathbf{^sT_c}
$$ 

## 3D Rotation Representations and DOF

All rotation representations must account for **3 Degrees of Freedom (DOF)**.

| Representation | Description | Initial free variables (Numbers) | Constraints | Final DOF |
| :--- | :--- | :--- | :--- | :--- |
| **Rotation Matrix** $\mathbf{R}$ | $3 \times 3$ matrix | 9 DOF | $\det(R)=1$, $R^T R = I$  | **3 DOF**  |
| **Euler Angles** | $(\theta_x, \theta_y, \theta_z)$  | 3 DOF | None | **3 DOF**  |
| **Quaternions** $\mathbf{q}$ | $(x, y, z, w)$ | 4 DOF | $\|\mathbf{q}\| = 1$ (One constraint) | **3 DOF**  |
| **Axis-Angle** | $(\mathbf{\hat{u}}, \theta_u)$ | 4 DOF | $\|\mathbf{\hat{u}}\| = 1$ (Unit vector constraint) | **3 DOF** |

## Roll-Pitch-Yaw to Rotation Matrix Conversion

For the Roll-Pitch-Yaw (RPY) sequence $R = R_x(\theta_x) R_y(\theta_y) R_z(\theta_z)$ , the individual rotation matrices are multiplied:

$$R = 
\begin{bmatrix}
c\theta_z & -s\theta_z & 0 \\
s\theta_z & c\theta_z & 0 \\
0 & 0 & 1
\end{bmatrix}
\begin{bmatrix}
c\theta_y & 0 & s\theta_y \\
0 & 1 & 0 \\
-s\theta_y & 0 & c\theta_y
\end{bmatrix}
\begin{bmatrix}
1 & 0 & 0 \\
0 & c\theta_x & -s\theta_x \\
0 & s\theta_x & c\theta_x
\end{bmatrix}
$$

This product yields the final $3 \times 3$ rotation matrix:

$$R = 
\begin{bmatrix}
r_{11} & r_{12} & r_{13} \\
r_{21} & r_{22} & r_{23} \\
r_{31} & r_{32} & r_{33}
\end{bmatrix}
$$

Where $c\theta = \cos\theta$ and $s\theta = \sin\theta$.


## Rotation Matrix to Roll-Pitch-Yaw Conversion

Given a rotation matrix $R$, the Euler angles $(\theta_x, \theta_y, \theta_z)$ (Roll, Pitch, Yaw) can be extracted:

1.  **Pitch ($\mathbf{\theta_y}$):**
    The pitch angle can be calculated from the $r_{31}$ element.
    $$- \sin(\theta_y) = r_{31}
    $$
    $$
    \theta_y = \sin^{-1}(-r_{31}) \in [-\frac{\pi}{2}, \frac{\pi}{2}]
    $$

2.  **Roll ($\mathbf{\theta_x}$):**
    The roll angle can be calculated using the $\arctan2$ function for quadrant-aware calculation:
    $$
    \theta_x = \arctan2(r_{32}, r_{33}) \in (-\pi, \pi]
    $$

3.  **Yaw ($\mathbf{\theta_z}$):**
    The yaw angle can be similarly calculated using $\arctan2$.
    $$
    \theta_z = \arctan2(r_{21}, r_{11})
    $$


## Gimbal Lock

**Gimbal Lock** is a significant problem encountered when using Euler angles.

It occurs when the **pitch angle** is $\theta_y = 90^{\circ} = \frac{\pi}{2}$.
* At this angle, $\cos(\theta_y) = 0$.
* The first and third rotation axes (Roll and Yaw axes) become aligned (swapped).
* This causes a loss of one Degree of Freedom (DOF), meaning two of the three rotation angles are redundant or indistinguishable.


## Axis-Angle Representation (Rodrigues' Formula)

The **Axis-angle representation** defines a rotation by a **unit vector** $\mathbf{\hat{k}} = [k_x, k_y, k_z]^T$ and a rotation **angle** $\theta$.
* The axis must be a unit vector: $||\mathbf{\hat{k}}|| = \sqrt{k_x^2+k_y^2+k_z^2}=1$.

The **Rodrigues' Formula** converts this representation into the rotation matrix $R$:
$$ 
R = I + \sin\theta [\mathbf{\hat{k}}_\times] + (1-\cos\theta) [\mathbf{\hat{k}}_\times]^2
$$ 

### Cross Product Matrix
The $[\mathbf{\hat{k}}_\times]$ notation represents the **Cross product matrix**:

$$ 
[\mathbf{k}_\times] = 
\begin{bmatrix}
0 & -k_z & k_y\\
k_z & 0 & -k_x\\
-k_y & k_x & 0
\end{bmatrix}
$$

The cross product of two vectors, $\mathbf{a} \times \mathbf{b}$, can be computed as a matrix multiplication using the cross-product matrix of $\mathbf{a}$:
$$
\mathbf{a} \times \mathbf{b} = [\mathbf{a}_\times]\mathbf{b}
$$

The matrix $A = [\mathbf{a}_\times]$ is a **Skew-symmetric matrix**, meaning $A^T = -A$.

### Applying the Axis-Angle Rotation to a Point
Applying the rotation $R$ to a point $\mathbf{p}$ is:

$$
R\mathbf{p} = \mathbf{p} + \sin\theta(\mathbf{\hat{k}}\times\mathbf{p}) + (1-\cos\theta)\mathbf{\hat{k}}\times(\mathbf{\hat{k}}\times\mathbf{p})
$$
