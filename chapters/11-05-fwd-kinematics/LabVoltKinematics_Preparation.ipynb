```python
def cross_prod_mat(k):
    return np.array([[0, -k[2], k[1]],
                     [k[2], 0, -k[0]],
                     [-k[1], k[0], 0]])

def rot3D(k, theta):
    k = np.asarray(k)
    k /= np.linalg.norm(k)
    K = cross_prod_mat(k)
    return np.eye(3) + np.sin(theta) * K + (1-np.cos(theta)) * K @ K

def rot_transform(R):
    T = np.eye(4)
    T[:3, :3] = R
    return T

def translation(xyz):
    T = np.eye(4)
    T[:3, 3] = xyz
    return T
```

```python
def dh_param_transform(d, theta, r, alpha):
    T_d_theta = np.array([
        [np.cos(theta), -np.sin(theta), 0, 0],
        [np.sin(theta),  np.cos(theta), 0, 0],
        [            0,              0, 1, d],
        [            0,              0, 0, 1],
    ])
    T_r_alpha = np.array([
        [1,             0,              0, r],
        [0, np.cos(alpha), -np.sin(alpha), 0],
        [0, np.sin(alpha),  np.cos(alpha), 0],
        [0,             0,              0, 1],
    ])
    
    return T_d_theta @ T_r_alpha
```

```python
import numpy as np
dh_param_table = np.array([
    [27.2, np.pi/180*10, 0, np.pi/2],
    [0, +np.pi/180*102.5, 19.2, 0],
    [0, -np.pi/180*54, 19.2, 0],
    [0, np.pi/2+np.pi/180*8, 0, np.pi/2],
    [0, 0, 0, 0],
])
dh_param_table_scaled_reset = np.array([
    [257, +np.pi/180*10, -10, 0],
    [0, +np.pi/180*102.5, 200, 0],
    [0, -np.pi/180*54, 190, 0],
    [3, np.pi/2+np.pi/180*8, 0, np.pi/2],
    [0, 0, 0, 0],
])
```

```python
def transform_inverse(T):
    Tinv = np.eye(4)
    R = T[:3,:3]
    Tinv[:3, :3] = R.T
    Tinv[:3, 3] = - R.T @ T[:3, 3]
    return Tinv
    
```

```python
def forward_kinematics(thetas, dh_param_table_scaled):
    T_0_from_i = [np.eye(4)]*(len(dh_param_table_scaled)+1)
    T_0_from_i[0] = np.eye(4)
    for i in range(1, 1+dh_param_table_scaled.shape[0]):
        d, theta, r, alpha = dh_param_table_scaled[i-1]
        theta += thetas[i-1]
        T_im1_from_i = dh_param_transform(d, theta, r, alpha)
        T_0_from_i[i] = T_0_from_i[i-1] @ T_im1_from_i
    return T_0_from_i
```

```python
T_0_from_i_reset = forward_kinematics([2*np.pi/3, -3*np.pi/6, -np.pi/6, 0., 0.], dh_param_table_scaled)

transformed_parts = []
xyzs = []
for i in range(len(parts)):
    transformed_parts.append(parts[i].copy().apply_transform(T_0_from_i_reset[i]))
    xyzs.append(
        trimesh.creation.axis(transform=T_0_from_i_reset[i],
                                  axis_length=200, axis_radius=1)
    )
scene = trimesh.scene.Scene(transformed_parts+xyzs)
scene.show()
```
